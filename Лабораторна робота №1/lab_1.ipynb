## Контрольні питання

1. Що таке Git Bash і як він пов’язаний із GitHub? Git Bash — консольний емулятор для Windows, який надає інтерфейс командного рядка для роботи з системою контролю версій Git. Інструмент дозволяє виконувати команди локально та синхронізувати код з віддаленими репозиторіями на платформі GitHub.

2. Які основні переваги використання GitHub для спільної роботи над проектами? Платформа надає зручні інструменти для контролю версій, відстеження помилок (issues), проведення код-рев'ю (pull requests) та управління завданнями. Розробники можуть паралельно працювати у різних гілках, уникаючи конфліктів у головному коді, та обговорювати зміни безпосередньо у прив'язці до конкретних рядків.

3. Які кроки потрібно виконати для створення нового репозиторію на GitHub та додавання змін до нього? Для ініціалізації та налаштування виконуються такі дії:
   - Натиснути кнопку «New» на сторінці профілю GitHub.
   - Вказати назву репозиторію та натиснути «Create repository».
   - У локальному терміналі ініціалізувати Git (команда `git init`).
   - Додати файли до індексу (`git add .`).
   - Зробити коміт (`git commit -m "назва коміту"`).
   - Встановити адресу віддаленого репозиторію (`git remote add origin <url>`).
   - Відправити код на сервер (`git push -u origin main`).

4. Які основні компоненти GitHub, що використовуються при роботі з репозиторіями та змінами в них? Основні складові включають:
   - Гілки (branches) для ізольованої розробки.
   - Коміти (commits) для фіксації змін.
   - Форки (forks) для копіювання чужих проектів.
   - Запити на злиття (pull requests) для пропонування своїх змін.
   - Розділ Issues для обговорення багів та планування нових функцій.

5. Яким чином ви можете відкрити запит на включення змін у GitHub і як такий механізм сприяє співпраці у розробці програмного забезпечення? Після відправки нової гілки на сервер потрібно перейти на сторінку репозиторію і натиснути кнопку «Compare & pull request». Далі додається опис запропонованих змін і натискається «Create pull request». Такий підхід дозволяє іншим учасникам команди переглянути запропонований код, залишити коментарі, провести тестування та безпечно інтегрувати нові функції у головну гілку без ризику зламати робочий проект.
